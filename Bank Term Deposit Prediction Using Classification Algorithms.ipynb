```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

bank_full=pd.read_csv('bank-full.csv', sep=';')
bank_full.head()
#bank_full.info()

features=['age', 'job', 'marital', 'education', 'balance', 'housing', 'loan', 'campaign', 'previous', 'poutcome']
X=bank_full[features]
y=bank_full['y']

X_train, X_test, y_train, y_test=train_test_split(X,y, test_size=0.2, random_state=1, stratify=y)

numerical_features = ['age', 'balance', 'campaign', 'previous']
categorical_features = ['job', 'marital', 'education', 'housing', 'loan', 'poutcome']

preprocessor=ColumnTransformer(transformers=[('num', StandardScaler(), numerical_features), ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)])

X_train_processed=preprocessor.fit_transform(X_train)
X_test_processed=preprocessor.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

LR_model=LogisticRegression()
LR=LR_model.fit(X_train_processed, y_train)
LR_pred=LR_model.predict(X_test_processed)
print("Logistic Regression:")
print("Accuracy:", accuracy_score(y_test, LR_pred))
print("Precision:", precision_score(y_test, LR_pred, pos_label='yes'))
print("Recall:", recall_score(y_test, LR_pred, pos_label='yes'))
print("F1 Score:", f1_score(y_test, LR_pred, pos_label='yes'))
print(confusion_matrix(y_test, LR_pred, labels=['no', 'yes']))
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_model=DecisionTreeClassifier()
tree=tree_model.fit(X_train_processed, y_train)
tree_pred=tree_model.predict(X_test_processed)
print("Decison Tree Classifier:")
print("Accuracy:", accuracy_score(y_test, tree_pred))
print("Precision:", precision_score(y_test, tree_pred, pos_label='yes'))
print("Recall:", recall_score(y_test, tree_pred, pos_label='yes'))
print("F1 Score:", f1_score(y_test, tree_pred, pos_label='yes'))
print(confusion_matrix(y_test, tree_pred, labels=['no', 'yes']))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_processed, y_train)
knn_pred=knn.predict(X_test_processed)
print("K-Nearest Neighbors:")
print("Accuracy:", accuracy_score(y_test, knn_pred))
print("Precision:", precision_score(y_test, knn_pred, pos_label='yes'))
print("Recall:", recall_score(y_test, knn_pred, pos_label='yes'))
print("F1 Score:", f1_score(y_test, knn_pred, pos_label='yes'))
print(confusion_matrix(y_test, knn_pred, labels=['no', 'yes']))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model=RandomForestClassifier()
rf_model.fit(X_train_processed, y_train)
rf_pred=rf_model.predict(X_test_processed)
print("Random Forest:")
print("Accuracy:", accuracy_score(y_test, rf_pred))
print("Precision:", precision_score(y_test, rf_pred, pos_label='yes'))
print("Recall:", recall_score(y_test, rf_pred, pos_label='yes'))
print("F1 Score:", f1_score(y_test, rf_pred, pos_label='yes'))
print(confusion_matrix(y_test, rf_pred, labels=['no', 'yes']))
```

```python
from sklearn.svm import LinearSVC

svm=LinearSVC()
svm.fit(X_train_processed, y_train)
svm_pred=svm.predict(X_test_processed)
print("Support Vector Machine:")
print("Accuracy:", accuracy_score(y_test, svm_pred))
print("Precision:", precision_score(y_test, svm_pred, pos_label='yes'))
print("Recall:", recall_score(y_test, svm_pred, pos_label='yes'))
print("F1 Score:", f1_score(y_test, svm_pred, pos_label='yes'))
print(confusion_matrix(y_test, svm_pred, labels=['no', 'yes']))
```
